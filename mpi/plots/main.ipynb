```python
import pandas as pd
```

```python
data = pd.read_csv(r"..\resultados_completo.csv")
data.sample(5)
data = data[['Experimento',	'N',	'K'	,'Processos'		,'SSE_Final'	,'Tempo_s']]
```

```python
dic = {
    "Escalabilidade":"Processos",
    "Variacao_N":"N",
    "Variacao_K":"K"
}

y_values_columns = ['SSE_Final', 'Tempo_s']
```

```python
list_all_columns = data.columns
y_values_columns_conjunto = set(y_values_columns)
x_files = [item for item in list_all_columns if item not in y_values_columns_conjunto]
x_files
```

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

for exp in list(dic.keys()):
    data_filtered = data.query(f"Experimento == '{exp}'")
    
    x_column = dic.get(exp)
    tabled_columns = [col for col in x_files if col != x_column]
    
    fig, ax1 = plt.subplots(figsize=(8, 6))
    ax2 = ax1.twinx()
    
    line1, = ax1.plot(
        data_filtered[x_column],
        data_filtered[y_values_columns[0]],
        color="blue",
        label=y_values_columns[0]
    )
    
    line2, = ax2.plot(
        data_filtered[x_column],
        data_filtered[y_values_columns[1]],
        color="red",
        label=y_values_columns[1]
    )
    
    ax1.set_xlabel(x_column)
    ax1.set_ylabel(y_values_columns[0], color="blue")
    ax2.set_ylabel(y_values_columns[1], color="red")
    plt.title(f"Experimento {exp}")
    
    ax1.legend(handles=[line1, line2], loc="upper left")
    
    fixed_values = [f"{col}: {data_filtered[col].iloc[0]}" for col in tabled_columns]
    fixed_text = "\n".join(fixed_values)
    
    ax1.text(
        0.95, 0.05, f"Valores fixos\n{fixed_text}",
        transform=ax1.transAxes,
        fontsize=9, va="bottom", ha="right",
        bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgray", alpha=0.7)
    )
    
    fig.tight_layout()
    fig.savefig(f"grafico_{exp}.png")
    plt.close(fig)
```
